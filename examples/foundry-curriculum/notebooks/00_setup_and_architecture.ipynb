## Architecture boundary

```text
notebook or packaged app
  -> logical model: foundry-chat
  -> selected aai-platform config
  -> Foundry project endpoint + model deployment
  -> Entra token and project RBAC
```

The endpoint and deployment vary by environment. Prompts, tools, and application logic do not. Never print the complete process environment or serialize the raw settings object.
